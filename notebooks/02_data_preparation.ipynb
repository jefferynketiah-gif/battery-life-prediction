```python
import pandas as pd
from sklearn.model_selection import KFold

# Updated to use the relative path matching your project structure
df = pd.read_csv('../data/processed/battery_summary.csv')

# Our feature set - everything we engineered, decided based on EDA findings
feature_cols = ["capacity_at_cycle_10", "capacity_at_cycle_100", 
                 "capacity_fade_10_to_100", "internal_resistance_at_100"]
target_col = "cycle_life"

X = df[feature_cols]
y = df[target_col]

kf = KFold(n_splits=5, shuffle=True, random_state=42)

for fold_num, (train_idx, test_idx) in enumerate(kf.split(X)):
    print(f"Fold {fold_num + 1}: {len(train_idx)} training cells, {len(test_idx)} testing cells")
```

We use 5-fold cross-validation instead of a single train/test split because with only 44 cells, one split's test set would be just 9 cells — small enough that the result could look good or bad mostly by chance, depending on which 9 cells happened to land there. By rotating through 5 different splits, every cell gets tested exactly once, and we get 5 separate performance scores instead of one. This lets us report both the average performance and how much it varies across folds – telling us not just "How good is the model?" but "How much can we actually trust that number?"
